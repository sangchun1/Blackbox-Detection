# 16 · Build Stage1 A7 + Stage3 v5-C submission

v5-C selection을 검증한 뒤 실행하는 최종 제출 ZIP builder입니다.
Stage1/2와 requirements.txt는 stage1_a7.zip에서 byte-lock합니다.


```python
from pathlib import Path
import subprocess
import shutil
import sys

REPO = Path("/content/Blackbox-Detection")
BRANCH = "stage3-sangchun"
REPO_URL = "https://github.com/sangchun1/Blackbox-Detection.git"

# Runtime에 같은 이름의 비-git 폴더가 있으면 제거
if REPO.exists() and not (REPO / ".git").is_dir():
    print("Removing stale non-git directory:", REPO)
    shutil.rmtree(REPO)

# Repo가 없으면 clone
if not (REPO / ".git").is_dir():
    subprocess.run(
        [
            "git",
            "clone",
            "--branch", BRANCH,
            "--single-branch",
            REPO_URL,
            str(REPO),
        ],
        check=True,
    )
else:
    subprocess.run(
        ["git", "-C", str(REPO), "fetch", "origin", BRANCH],
        check=True,
    )
    subprocess.run(
        ["git", "-C", str(REPO), "checkout", BRANCH],
        check=True,
    )
    subprocess.run(
        ["git", "-C", str(REPO), "pull", "--ff-only", "origin", BRANCH],
        check=True,
    )

SCRIPT = REPO / "scripts/stage3/build_stage1a7_stage3_v5c_submission.py"

head = subprocess.run(
    ["git", "-C", str(REPO), "rev-parse", "--short", "HEAD"],
    check=True,
    capture_output=True,
    text=True,
).stdout.strip()

print("repo HEAD :", head)
print("builder   :", SCRIPT)
print("exists    :", SCRIPT.is_file())

assert (REPO / ".git").is_dir()
assert SCRIPT.is_file(), SCRIPT

print("REPO SETUP PASS")
```
